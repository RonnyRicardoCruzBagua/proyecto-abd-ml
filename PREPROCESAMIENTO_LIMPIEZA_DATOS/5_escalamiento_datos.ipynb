# Escalamiento de Variables Numéricas

En esta etapa se realiza la estandarización de las variables numéricas mediante StandardScaler, con el objetivo de garantizar que todas las características del conjunto de datos se encuentren en una escala comparable para los algoritmos de Machine Learning.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("7_dataset_transformado.csv")
```

## Identificación de Variables Numéricas

```python
columnas_numericas = df.select_dtypes(
    include=['int64', 'float64']
).columns

print("Número de variables numéricas:", len(columnas_numericas))

columnas_numericas
```

## Aplicación de StandardScaler

```python
scaler = StandardScaler()

df[columnas_numericas] = scaler.fit_transform(
    df[columnas_numericas]
)
```

## Vista Previa del Dataset Escalado

```python
df.head()
```

## Validación del Escalamiento

```python
df.describe()
```

## Resumen del Proceso

```python
resumen = pd.DataFrame({
    'Métrica': [
        'Registros',
        'Columnas',
        'Variables Escaladas'
    ],
    'Valor': [
        df.shape[0],
        df.shape[1],
        len(columnas_numericas)
    ]
})

resumen
```

## Exportación del Dataset Final

Una vez finalizado el proceso de preprocesamiento, transformación y escalamiento, se genera el conjunto de datos preparado para las siguientes fases del proyecto.

```python
df.to_csv(
    "8_dataset_preparado_ml.csv",
    index=False
)
```

## Conclusión

Se realizó el escalamiento de las variables numéricas utilizando la técnica StandardScaler. Como resultado, todas las variables quedaron en una escala homogénea, permitiendo mejorar el desempeño y estabilidad de futuros modelos de Machine Learning.

El archivo generado constituye la versión final del dataset procesado para las etapas de entrenamiento y evaluación de modelos predictivos.
