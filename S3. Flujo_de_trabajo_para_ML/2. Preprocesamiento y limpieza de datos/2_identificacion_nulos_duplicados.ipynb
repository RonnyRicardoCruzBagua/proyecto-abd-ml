# Análisis de Calidad de Datos

Se realiza una evaluación de la calidad del conjunto de datos mediante la identificación de valores faltantes y registros duplicados.


```python
import pandas as pd

df = pd.read_csv('../S0. Database/hotel_bookings.csv')
```

```python
calidad_datos = pd.DataFrame({
    'tipo_dato': df.dtypes,
    'nulos': df.isnull().sum(),
    'porcentaje_nulos': round((df.isnull().sum()/len(df))*100,2),
    'valores_unicos': df.nunique()
})

calidad_datos.sort_values(
    by='porcentaje_nulos',
    ascending=False
)
```

```python
duplicados = df.duplicated().sum()

print(f'Registros duplicados encontrados: {duplicados}')
```
