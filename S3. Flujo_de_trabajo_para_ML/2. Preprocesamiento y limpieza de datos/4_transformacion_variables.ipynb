# **Transformación de Variables Categóricas**

En esta etapa se convierten las variables categóricas a valores numéricos mediante técnicas de codificación, permitiendo su utilización en algoritmos de Machine Learning.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("6_dataset_limpio.csv")
```

## Identificación de Variables Categóricas

```python
columnas_categoricas = df.select_dtypes(
    include='object'
).columns

columnas_categoricas
```

## Aplicación de Label Encoding

```python
encoder = LabelEncoder()

for columna in columnas_categoricas:
    df[columna] = encoder.fit_transform(
        df[columna].astype(str)
    )
```
